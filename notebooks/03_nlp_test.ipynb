```python
# Cellule 1 : Test du NLP
import sys
import os
import torch

# Path setup
sys.path.append(os.path.abspath(os.path.join('..', 'src')))

from neural_quant.nlp.extractor import CausalEventExtractor

def test_nlp_engine():
    # 1. Initialisation (Télécharge le modèle ~1.5GB la première fois)
    print("Initialisation du NLP Engine...")
    extractor = CausalEventExtractor()
    
    # 2. Fake News Dataset (Mélange de bruits et de signaux forts)
    news_feed = [
        "Copper mine in Chile halts production due to labor strike", # Supply Shock
        "Goldman Sachs predicts steady growth for tech sector",     # Market Noise (pour les métaux)
        "Russia announces new export bans on Nickel",               # Geopolitics + Supply
        "Local football team wins championship",                    # Noise total
        "Massive earthquake hits Peru mining region",               # Supply Shock naturel
    ]
    
    print("\n--- Analyse des News ---")
    
    # 3. Traitement
    z_signals = extractor.process_news_batch(news_feed)
    
    # 4. Affichage des résultats
    for news, score in zip(news_feed, z_signals):
        # On récupère la valeur scalaire
        val = score.item()
        
        # Interprétation visuelle
        bar = "█" * int(val * 2) # Barre de chargement
        status = "CRISIS EVENT" if val > 2.0 else "Noise/Calm"
        
        print(f"\nNews: {news}")
        print(f"Signal Z_t: {val:.4f}  [{bar}]  -> {status}")

if __name__ == "__main__":
    test_nlp_engine()
```
